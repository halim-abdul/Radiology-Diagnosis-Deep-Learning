```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve
from radiology.training.metrics import binary_metrics
y=np.array([0,0,1,1,0,1]);p=np.array([.1,.35,.45,.8,.2,.9]);print(binary_metrics(y,p))
fpr,tpr,_=roc_curve(y,p);plt.plot(fpr,tpr,label='Toy ROC');plt.plot([0,1],[0,1],'--');plt.legend();
```
