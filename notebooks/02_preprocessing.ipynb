# Intensity normalization and contrast histogram
**Research-only synthetic demonstration. No clinical claims.**
Run after `pip install -e '.[test]'`. Images are generated by plotting cells.


```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
a=rng.normal(100,20,(128,128));n=(a-a.min())/(a.max()-a.min())
fig,axes=plt.subplots(1,2);axes[0].imshow(n,cmap='gray');axes[1].hist(n.ravel(),bins=30);
```
